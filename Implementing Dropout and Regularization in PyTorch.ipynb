```python
import torch
from torch import nn
```

```python
net = nn.Sequential(nn.Linear(2,4),nn.ReLU(),nn.Dropout(0.5),nn.Linear(4,1))
print(net)
```

```python
for p in net[0].parameters():
    print(p.data)  # parameter of different layers of the model.
    print(p.abs().sum())   # Every layers we can sum 
for i,p in enumerate(net[0].parameters()):
    print(i,p.abs().sum())
```

```python
layers_all=[]
for i,p in enumerate(net[0].parameters()):
    print(i,p.abs().sum())
    layers_all.append(p.abs().sum())
print(sum(layers_all))
```

# Adding L1 and L2 regularisation 

```python
import torch.optim as optim
optimizer = optim.SGD(net.parameters(), lr=0.01, weight_decay=0.01)
```

```python
l1_lambda=0.01

loss_function = nn.MSELoss()
X = torch.tensor([[ 0.1247,  0.6003],
        [-0.3116,  0.0329]])
target = torch.tensor([[1.0], [0.0]])  

output=net(X)
loss=loss_function(output, target)  #Forward pass
l1_penalty = l1_lambda * sum(param.abs().sum() for param in net.parameters())
loss+=l1_penalty
print(loss)
```

# Training with Dropout and Regularization

```python
net. train() # Set the network to training mode
for epoch in range(100):
    output=net(X)
    loss = loss_function(output, target)

    l1_penalty=l1_lambda * sum(param.abs().sum() for param in net.parameters())
    loss+=l1_penalty

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if epoch % 10 == 0:
        print(f"Epoch{epoch}, loss:{loss.item()}")
```

```python
net. train() # Set the network to training mode
l2_lambda=0.01
for epoch in range(100):
    output=net(X)
    loss = loss_function(output, target)

    l2_penalty=l2_lambda * sum((param**2).sum() for param in net.parameters())
    loss+=l2_penalty

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if epoch % 10 == 0:
        print(f"Epoch{epoch}, loss:{loss.item()}")
```
